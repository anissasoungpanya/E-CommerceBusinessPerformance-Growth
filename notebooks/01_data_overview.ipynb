# Orders

```python
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
orders.head()
```

```python
print("Orders dataset overview")

print("Dataset Shape:", orders.shape)
print("\nColumn Names:")
print(orders.columns.tolist())

print("\nData Types:")
print(orders.dtypes)

print("\nMissing Values:")
print(orders.isnull().sum())

missing_percentage = (orders.isnull().sum() / len(orders)) * 100

print("\nMissing Value Percentage:")
print(missing_percentage.round(2))

print("\nDuplicate Rows:", orders.duplicated().sum())

print("\nDuplicate Orders:", orders['order_id'].duplicated().sum())

print("\nOrder Statuses:")
print(orders["order_status"].unique())

order_status_counts = orders["order_status"].value_counts()

print("\nOrder Status Distribution:")
print(order_status_counts)

order_status_percentage = orders["order_status"].value_counts(normalize=True) * 100

print("\nOrder Status Percentage:")
print(order_status_percentage.round(2))

purchase_dates = pd.to_datetime(orders["order_purchase_timestamp"])

print("\nEarliest Purchase Date:", purchase_dates.min())
print("Latest Purchase Date:", purchase_dates.max())
print("Total Days Covered:", (purchase_dates.max() - purchase_dates.min()).days)
```

# Payments

```python
payments = pd.read_csv('../data/raw/olist_order_payments_dataset.csv')
payments.head()
```

```python
print("Payments Dataset Overview")

print("Dataset Shape:", payments.shape)
print("\nColumn Names:")
print(payments.columns.tolist())

print("\nData Types:")
print(payments.dtypes)

print("\nMissing Values:")
print(payments.isnull().sum())

print("\nDuplicate Rows:", payments.duplicated().sum())

print("\nTotal Payment Value:", payments["payment_value"].sum())

payment_summary = pd.DataFrame({
    "Count": payments["payment_type"].value_counts(),
    "Percentage": payments["payment_type"].value_counts(normalize=True) * 100
})

payment_summary = payment_summary.sort_values("Count", ascending=False)

print('\n', payment_summary.round(2))

print("\nMax and Min Payment Installments:")
print(payments["payment_installments"].max(), payments["payment_installments"].min())
```
