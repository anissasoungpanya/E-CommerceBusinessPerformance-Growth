# Product Category Translation Data Cleaning

Problems to investigate:
- Duplicate category names or translations
- Any whitespace

```python
import pandas as pd 

translate_raw = pd.read_csv('../data/raw/product_category_name_translation.csv')
translate = translate_raw.copy()

print("Original Shape:", translate_raw.shape)
print("Working Shape:", translate.shape)
```

```python
# Check for duplicate category names or translations

duplicate_names = translate["product_category_name"].duplicated().sum()

print("Duplicate category names:", duplicate_names)

duplicate_translations = translate["product_category_name_english"].duplicated().sum()

print("Duplicate translated names:", duplicate_translations)
```

```python
# Check for any whitespace

string_columns = translate.select_dtypes(include=["object", "str"]).columns

for column in string_columns:
    count = translate[column].str.contains(
        r"^\s|\s$", na=False
    ).sum()

    print(f"{column}: {count} values with whitespace")

translate.columns = (
    translate.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(translate.columns.tolist())
```

```python
translate.to_csv(
    "../data/processed/product_category_name_translation_cleaned.csv",
    index=False
)
```
